# Internal Pipe Ultrasonic Sim — quick exploration

Run a single-angle pulse-echo simulation and inspect NPZ outputs.
For BC segment multi-year studies, use `well-array-sim bc run` from the shell.

```python
from pathlib import Path
import sys

REPO = Path.cwd().resolve()
if not (REPO / "src" / "well_array_sim").exists():
    REPO = REPO.parent
sys.path.insert(0, str(REPO / "src"))

from well_array_sim.cli import _run_sim_cli

out = REPO / "outputs" / "notebook_demo"
_run_sim_cli([
    "--scenario", "scenarios/internal_pipe_default.yaml",
    "--angle-deg", "45",
    "--out", str(out),
])
print(f"Wrote plots and {out}_pulse_echo.npz")
```

```python
import numpy as np
from pathlib import Path
from IPython.display import Image, display

REPO = Path.cwd().resolve()
if not (REPO / "outputs").exists():
    REPO = REPO.parent

prefix = REPO / "outputs" / "notebook_demo"
for name in ("packet_scene", "pulse_echo", "saft_profile"):
    path = Path(f"{prefix}_{name}.png")
    if path.exists():
        display(Image(filename=str(path)))

npz = np.load(f"{prefix}_pulse_echo.npz")
print("NPZ keys:", list(npz.keys()))
print("inferred distance (mm):", float(npz["inferred_distance_m"]) * 1000)
```
